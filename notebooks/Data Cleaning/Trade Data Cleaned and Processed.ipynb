```python
# imports
import pandas as pd
import warnings
import pycountry_convert as pc
warnings.filterwarnings("ignore", category=FutureWarning)
```

```python
# Load the data
trade_data = pd.read_excel('../Original Data/TradeData.xlsx')

# View the data
trade_data.head(10)
```

```python
trade_data.info()
```

```python
columns_to_keep = [
    'refYear', 'reporterDesc',
    'flowDesc', 'partnerDesc', 'cmdCode', 'qty',
    'primaryValue', 'reporterISO', 'partnerISO'
]

# New data frame with only the selected columns
trade_data = trade_data[columns_to_keep]
```

```python
# changes the comodity codes into more readable format
trade_data['cmdCode'] = trade_data['cmdCode'].replace({2709: 'Oil', 2701: 'Coal', 271111: 'Gas'})

# checks the dataframe updated
print(trade_data.head(20))
```

```python
# Checks for percent of any missing values
percentage_missing = trade_data.isnull().sum() * 100 / len(trade_data)
missing_value_df = pd.DataFrame({'percent_missing': percentage_missing})

print(missing_value_df)
```

```python
# Renames the columns
trade_data = trade_data.rename(columns={"refYear": "year", "reporterDesc": "exporter", "flowDesc": "flow", "partnerDesc": "importer", "cmdCode": "resource",
                    "qty": "weight (kg)", "primaryValue": "value (US$)", "reporterISO": "exporterISO", "partnerISO": "importerISO"})
# Check column names updated
trade_data.head()
```

```python
# Code adapted from [https://stackoverflow.com/questions/55910004/get-continent-name-from-country-using-pycountry]
def country_to_continent(country_name):
    try:
        country_alpha2 = pc.country_name_to_country_alpha2(country_name)
        country_continent_code = pc.country_alpha2_to_continent_code(country_alpha2)
        country_continent_name = pc.convert_continent_code_to_continent_name(country_continent_code)
        return country_continent_name
    except (KeyError, AttributeError):
        return None


trade_data["exporter continent"] = trade_data['exporterISO'].apply(country_to_continent)
trade_data["importer continent"] = trade_data['importerISO'].apply(country_to_continent)
```

```python
trade_data = trade_data.replace("Russian Federation", "Russia")
```

```python
# checks for any missing values in partnerISO and reporterISO field
missing_partner = trade_data['importerISO'].isna().sum()
missing_reporter = trade_data['exporterISO'].isna().sum()
missing_continent_partner = trade_data['exporter continent'].isna().sum()
missing_continent_reporter = trade_data['importer continent'].isna().sum()
print(f"{missing_partner} for partnerISO")
print(f"{missing_reporter} for reporterISO")
print(f"{missing_continent_partner} for partner continent")
print(f"{missing_continent_reporter} for reporter continent")
```

```python
missing_exporter_isos = trade_data[trade_data['exporter continent'].isna()]['exporter'].unique()
missing_importer_isos = trade_data[trade_data['importer continent'].isna()]['importer'].unique()
# prints the unique ISO codes with missing continents
print(f"ReporterISO with missing continents: {missing_exporter_isos}")

print(f"PartnerISO with missing continents: {missing_importer_isos}")
```

```python

# keeps the rows where exporter and importer continent is not na
trade_data = trade_data[trade_data['exporter continent'].notna()]
trade_data = trade_data[trade_data['importer continent'].notna()]
missing_continent_partner = trade_data['exporter continent'].isna().sum()
missing_continent_reporter = trade_data['importer continent'].isna().sum()
print(f"{missing_continent_partner} for partner continent")
print(f"{missing_continent_reporter} for reporter continent")
```
