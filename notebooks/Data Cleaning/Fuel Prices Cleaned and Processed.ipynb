```python
# imports
import pandas as pd
import warnings
import pycountry_convert as PC
import pycountry
warnings.filterwarnings("ignore", category=FutureWarning)
# load the data
fuel_data = pd.read_excel('../Original Data/Fuel Prices Europe.xlsx', sheet_name=1)
fuel_data
```

```python
# country abbr map to change value names in country later
country_abbreviation = {
    "BE": "Belgium",
    "DE": "Germany",
    "DK": "Denmark",
    "LU": "Luxembourg",
    "NL": "Netherlands",
    "IE": "Ireland",
    "ES": "Spain",
    "PT": "Portugal",
    "IT": "Italy",
    "FR": "France",
    "GR": "Greece",
    "AT": "Austria",
    "SE": "Sweden",
    "FI": "Finland",
    "SI": "Slovenia",
    "LT": "Lithuania",
    "MT": "Malta",
    "CY": "Cyprus",
    "EE": "Estonia",
    "LV": "Latvia",
    "PL": "Poland",
    "CZ": "Czechia",
    "SK": "Slovakia",
    "HU": "Hungary",
    "BG": "Bulgaria",
    "RO": "Romania",
    "HR": "Croatia",
    "UK": "United Kingdom"
}
```

```python
# Initial Data Proce
# Rename column to dates
fuel_data.rename(columns={'Consumer prices of petroleum products net of duties and taxes': 'Date'}, inplace=True)
# Drop rows containing NA/text values
fuel_data.drop([0, 1, 1029, 1028, 1006, 1007, 1008,1009,
               1010,1011,1012,1013, 1014, 1015, 1016, 1017,1018,1019,
                1020,1021,1022,1023,1024,1025,1026,1027], axis=0, inplace=True)
# convert date column to datetime format 
fuel_data["Date"] = pd.to_datetime(fuel_data["Date"])
# filter to only keep dates larger or equal to than 2017
fuel_data = fuel_data[fuel_data["Date"] >= "2017-01-02"]
# create year column and add the years from date column
fuel_data["Year"] = fuel_data["Date"].dt.year
# drops the date column as now we are using year column
fuel_data.drop("Date", axis=1, inplace=True)
```

```python
# Melting the data wide to long
# https://www.youtube.com/watch?v=F_N6oa1bksE&ab_channel=SelvaPrabhakaran%28ML%2B%
# Changes the data from wide to long format with Year as the ID and prices of petrol and diesel as values
# values (prices in this case) get stored in prices and value_vars get stored in Country_Type
fuel_melted = fuel_data.melt( 
                    id_vars=['Year'],
                    value_vars=[  
                        "AT_price_wo_tax_euro95", "BE_price_wo_tax_euro95", "BG_price_wo_tax_euro95", "CY_price_wo_tax_euro95",
                        "CZ_price_wo_tax_euro95", "DE_price_wo_tax_euro95", "DK_price_wo_tax_euro95", "EE_price_wo_tax_euro95",
                        "ES_price_wo_tax_euro95", "FI_price_wo_tax_euro95", "FR_price_wo_tax_euro95", "GR_price_wo_tax_euro95", 
                        "HR_price_wo_tax_euro95", "HU_price_wo_tax_euro95", "IE_price_wo_tax_euro95", "IT_price_wo_tax_euro95",
                        "LT_price_wo_tax_euro95", "LU_price_wo_tax_euro95", "LV_price_wo_tax_euro95", "MT_price_wo_tax_euro95",
                        "NL_price_wo_tax_euro95", "PL_price_wo_tax_euro95", "PT_price_wo_tax_euro95", "RO_price_wo_tax_euro95",
                        "SE_price_wo_tax_euro95", "SI_price_wo_tax_euro95", "SK_price_wo_tax_euro95", "UK_price_wo_tax_euro95",
                        "AT_price_wo_tax_diesel", "BE_price_wo_tax_diesel", "BG_price_wo_tax_diesel", "CY_price_wo_tax_diesel",
                        "CZ_price_wo_tax_diesel", "DE_price_wo_tax_diesel", "DK_price_wo_tax_diesel", "EE_price_wo_tax_diesel",
                        "ES_price_wo_tax_diesel", "FI_price_wo_tax_diesel", "FR_price_wo_tax_diesel", "GR_price_wo_tax_diesel", 
                        "HR_price_wo_tax_diesel", "HU_price_wo_tax_diesel", "IE_price_wo_tax_diesel", "IT_price_wo_tax_diesel",
                        "LT_price_wo_tax_diesel", "LU_price_wo_tax_diesel", "LV_price_wo_tax_diesel", "MT_price_wo_tax_diesel",
                        "NL_price_wo_tax_diesel", "PL_price_wo_tax_diesel", "PT_price_wo_tax_diesel", "RO_price_wo_tax_diesel",
                        "SE_price_wo_tax_diesel", "SI_price_wo_tax_diesel", "SK_price_wo_tax_diesel", "UK_price_wo_tax_diesel",
                    ],
                    var_name='Country_Type',  
                    value_name='Price')
```

```python
fuel_melted
```

```python
# https://stackoverflow.com/questions/37333299/splitting-a-pandas-dataframe-column-by-delimiter
# Splits the data by delimiter (underscore) and creates a new column to split the data into.
# .str[0] takes the first element after splitting by _ (country in this case) and adds it to the Country column
fuel_melted['Country'] = fuel_melted['Country_Type'].str.split('_').str[0]
# .str[4] takes the fith element after splitting by _ (fuel in this case) and adds it to the Fuel column
fuel_melted['Fuel'] = fuel_melted['Country_Type'].str.split('_').str[4]
# deletes the unwanted columns
fuel_melted.drop(['Country_Type'],axis=1, inplace=True)
# Renames the fuels to more readable format
fuel_melted['Fuel'] = fuel_melted['Fuel'].str.replace("euro95", "Petrol")
fuel_melted['Fuel'] = fuel_melted['Fuel'].str.replace("diesel", "Diesel")
# maps the country abbreviations to country names
fuel_melted['Country'] = fuel_melted['Country'].map(country_abbreviation)

fuel_melted
```

```python
# Referenced from https://stackoverflow.com/questions/47919846/using-pycountry-to-check-for-name-common-name-official-name
import pycountry
def get_country_code(name):
    for co in list(pycountry.countries):
        if name in co.name:
            return co.alpha_3
    return None

fuel_melted["countryISO"] = fuel_melted["Country"].apply(get_country_code)
```

```python
# checks for any missing values in countryISO field
missing = fuel_melted['countryISO'].isna().sum()
print(f"{missing}")
```

```python
fuel_melted.dropna(axis=0, inplace=True)
```
